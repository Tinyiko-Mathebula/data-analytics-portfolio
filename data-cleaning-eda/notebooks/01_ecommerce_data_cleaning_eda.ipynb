## 4. Duplicate Records Investigation

### Duplicate Finding

The dataset contains 5,268 exact duplicate rows.

I confirmed that duplicate records can contain identical values across all transaction fields, including InvoiceNo, StockCode, Description, Quantity, InvoiceDate, UnitPrice, CustomerID, and Country.

InvoiceNo is not treated as a unique identifier because a single invoice can contain multiple product lines.

For the cleaning process, exact duplicate transaction rows will be removed while retaining the first occurrence.

### Missing CustomerID Finding

Approximately 25.16% of transactions have a missing CustomerID.

A missing CustomerID does not necessarily mean that the transaction is invalid. The transaction may still contain useful information such as the product, quantity, price, date, and country.

Therefore, CustomerID values will not be filled with invented values or removed solely because they are missing.

For customer-level analysis, transactions without a CustomerID may need to be excluded from analyses that specifically require customer identification.

### Missing Description Finding

There are 1,454 transactions with a missing Description.

Further investigation showed that all 1,454 of these records have a UnitPrice of 0.0.

Because the missing descriptions are associated with zero-priced records, they appear to represent non-standard transaction or adjustment records rather than normal product sales.

The original values will be preserved, but these records will be considered when defining the analysis dataset for sales and revenue calculations.

### Date Finding

The InvoiceDate column was originally stored as text and was converted to a datetime data type.

The transaction data covers the period from 1 December 2010 to 9 December 2011.

Converting InvoiceDate to datetime will allow time-based analysis such as daily, monthly, and yearly transaction trends.

### Zero Unit Price Finding

There are 2,515 transactions with a UnitPrice of 0.00.

Investigation showed that 1,056 of these records contain a Description. Several descriptions indicate non-standard operational transactions, including damages, adjustments, items found, items thrown away, and unsaleable or destroyed items.

Therefore, zero-priced records will not automatically be deleted from the raw dataset. For sales and revenue analysis, they will be treated as non-revenue transactions unless there is evidence that they represent a valid sale.

### Negative Unit Price Finding

There are 2 transactions with a negative UnitPrice.

Both records have the description "Adjust bad debt" and a UnitPrice of -11,062.06. They also use StockCode "B" and have invoice numbers beginning with "A".

These records appear to represent accounting adjustments rather than normal product sales.

The records will be preserved in the raw dataset but excluded from standard sales and revenue calculations.

### Invoice Prefix Finding

Invoice numbers beginning with "C" account for 9,251 rows.

A validation check found that there are no "C" invoice records with positive quantities. All "C" invoice records therefore have negative quantities.

This provides evidence that "C" invoice records represent cancellation or return-related transactions and should be separated from standard sales when calculating sales performance and revenue.

### A Invoice Finding

There are 3 invoice records beginning with "A".

All three records have StockCode "B" and the description "Adjust bad debt". They represent accounting adjustments rather than ordinary product sales.

The three records consist of one positive adjustment and two negative adjustments.

These records will be preserved in the dataset but excluded from standard product sales and revenue calculations.

### StockCode Finding

The dataset contains 4,070 unique StockCodes.

A total of 112 StockCodes have no product description anywhere in the dataset. Further investigation showed that all transactions associated with these 112 StockCodes have a UnitPrice of 0.00.

These records will be preserved in the raw dataset but treated as non-standard records when defining the sales analysis dataset.

### Quantity Outlier Finding

The quantity field contains extreme values, including both very large positive quantities and large negative quantities.

Investigation showed that some large positive quantities are associated with valid product descriptions and positive unit prices, suggesting that they may represent legitimate bulk transactions.

Large negative quantities are also present in cancellation or return-related invoices.

Therefore, extreme quantities will not be removed solely because of their magnitude. They will be retained unless further business validation indicates that they are erroneous.

### Country Formatting Finding

The dataset contains 38 unique country values.

A check for leading and trailing spaces produced the same number of unique values before and after applying `strip()`.

This indicates that leading or trailing whitespace is not creating duplicate country categories.

No country values will be changed at this stage.

### Description Formatting Finding

The description field contains capitalization inconsistencies. The dataset has 4,223 unique original descriptions, compared with 4,206 unique descriptions after converting the text to uppercase.

This indicates that some descriptions differ only in capitalization.

To improve consistency for analysis, a new `description_clean` column was created using uppercase text. The original `description` column was preserved so that the source values remain available for reference.

### Stock Code Formatting Finding

The dataset contains 4,070 unique original StockCodes.

After converting StockCodes to uppercase, the number of unique codes decreases to 3,958. This indicates that some product codes appear with inconsistent capitalization.

For example, the same code appears as `15056N` and `15056n`.

A new `stock_code_clean` column was created using uppercase text. The original `stock_code` column was preserved so that the source values remain available for reference.

The cleaned stock code will be used for product-level analysis.

### Customer ID Finding

The dataset contains 4,372 unique Customer IDs.

The `customer_id` column was stored as a floating-point data type because it contains missing values. Validation showed that all non-missing Customer IDs are whole numbers.

A new `customer_id_clean` column was created using the nullable integer (`Int64`) data type. The original `customer_id` column was preserved for reference.

Missing Customer IDs were not replaced with invented values because the actual customer identity cannot be reliably determined from the available data.

### Sales Analysis Dataset Definition

For sales and revenue analysis, a transaction is treated as a standard sale when:

- Quantity is greater than 0.
- Unit price is greater than 0.
- Invoice number does not start with `A`, which identifies accounting adjustment records.
- Invoice number does not start with `C`, which identifies cancellation or return-related records.

Using these rules results in 524,877 standard sales transaction rows.

Non-standard records are retained in the cleaned dataset but excluded from standard sales and revenue calculations.

### Monthly Sales Trend Finding

Month-over-month sales changes show that sales fluctuated throughout the analysis period.

The largest increases occurred in September 2011 (+41.47%) and March 2011 (+37.06%), while November 2011 recorded a 30.63% increase compared with October.

The largest decline occurred in December 2011 (-57.59%). However, December 2011 contains data only through 9 December, so it is a partial month and should not be directly compared with complete months.

Overall, the monthly results show periods of both growth and decline, with sales reaching their highest monthly value in November 2011 at approximately 1.50 million.

### Extreme Product Quantity Finding

The product `PAPER CRAFT , LITTLE BIRDIE` contains an unusually large quantity of 80,995 units in a single transaction.

The transaction has a valid stock code, positive unit price, identified customer, and positive sales value of 168,469.60.

Because there is insufficient evidence to classify the transaction as an error, it will be retained in the cleaned dataset. However, it will be treated as an extreme outlier and considered when interpreting product-level sales results.

### Transaction Volume Finding

Transaction volume increased substantially toward the end of the analysis period.

November 2011 recorded the highest transaction volume at 82,004 transaction lines, followed by October with 58,537 and September with 48,900.

This increase in transaction volume is consistent with the strong sales growth observed from September to November 2011.

December 2011 contains only partial-month data through 9 December and should therefore not be compared directly with the complete months.

### Transaction Value Finding

The overall average transaction value was 20.25.

November 2011 recorded the highest total sales value and the highest transaction volume, but its average transaction value was 18.34, below the overall average.

This indicates that the increase in November sales was driven primarily by higher transaction volume rather than unusually high average transaction values.

August and September recorded average transaction values above 21, while December recorded 25.68. However, December contains only partial-month data through 9 December and should therefore be interpreted with caution.

### Country Transaction Value Finding

The United Kingdom generated the highest total sales value and had the largest transaction volume, but its average transaction value was relatively low at 18.22.

Several international markets had substantially higher average transaction values. The Netherlands recorded an average of 122.26, Australia 117.04, and Japan 116.56.

These results show that total sales value and average transaction value provide different perspectives on market performance. The United Kingdom's sales were driven largely by transaction volume, while some smaller international markets generated higher value per transaction line.

The results should be interpreted carefully because transaction volume counts individual product lines rather than unique invoices.

### Customer Sales Finding

The customer-level analysis shows that sales are concentrated among a small group of high-value customers.

Customer 14646 generated the highest identified customer sales value at 280,206.02, followed by customers 18102 and 17450.

This indicates that a relatively small number of customers contributed substantial sales value during the analysis period.

These results are based only on transactions with an identified CustomerID. Transactions with missing CustomerID values were excluded from this customer-level analysis.

### Quantity and Sales Value Finding

The correlation between quantity and sales value is 0.908838, indicating a strong positive relationship between the two variables.

Higher quantities are generally associated with higher sales values in the dataset.

However, correlation does not imply causation. Sales value is calculated from quantity multiplied by unit price, so quantity is mathematically one of the components that determines sales value.

The scatter plot also shows the presence of extreme observations, which should be considered when interpreting the relationship.

### Customer Concentration Finding

The top 10 identified customers generated approximately 17.30% of total sales value among transactions with an identified CustomerID.

This indicates that a relatively small group of customers contributed a meaningful share of identified customer sales.

The result highlights the potential importance of customer retention and high-value customer management.

The calculation excludes transactions with missing CustomerID values because those transactions cannot be reliably attributed to individual customers.

# E-Commerce Transactions: Data Cleaning & EDA Insight Report

**Dataset:** Online Retail Transactions  
**Analysis Period:** 1 December 2010 – 9 December 2011  
**Tools:** Python, Pandas, NumPy, Matplotlib, Seaborn

## 1. Business Objective

The objective of this analysis was to clean and explore an e-commerce transaction dataset and identify patterns that could support business decision-making.

The analysis focused on sales performance, product performance, customer contribution, geographic markets, transaction trends, and data quality issues.

The analysis also aimed to identify unusual or non-standard records and define a reliable dataset for standard sales and revenue analysis.

## 2. Data Preparation & Cleaning

The original dataset contained 541,909 transaction rows and 8 columns.

The cleaning process included:

- Removing 5,268 exact duplicate rows.
- Converting `InvoiceDate` from text to a datetime format.
- Standardising column names and selected text fields.
- Investigating missing CustomerID and Description values.
- Investigating negative quantities, zero/negative unit prices, and unusual invoice prefixes.
- Separating cancellations, accounting adjustments, and other non-standard records from standard sales analysis.
- Creating `sales_value` as `quantity × unit_price`.

After cleaning and applying the standard sales criteria, 524,877 transaction lines were retained for standard sales and revenue analysis.

## 4. Key Business Insights

### 1. Sales increased strongly toward the end of the analysis period
Monthly sales reached approximately 1.50 million in November 2011, the highest complete-month sales value in the dataset. This was accompanied by the highest transaction-line volume of 82,004. The increase therefore appears to have been driven primarily by higher transaction volume rather than a higher average transaction line value.

### 2. The United Kingdom was the dominant market
The United Kingdom generated approximately 8.73 million in product sales, substantially more than the other countries analysed. This indicates a strong concentration of sales in the domestic market, while international markets such as the Netherlands, EIRE, Germany and France contributed smaller shares.

### 3. A small group of customers contributed a meaningful share of sales
The top 10 identified customers generated approximately 17.30% of sales attributed to customers with an identified CustomerID. This indicates that high-value customers represent an important component of the business's customer revenue base.

### 4. Product sales were concentrated among several high-performing products
REGENCY CAKESTAND 3 TIER generated the highest sales value among the clearly identified products at approximately 174,156.54. Other strong performers included WHITE HANGING HEART T-LIGHT HOLDER, PARTY BUNTING and JUMBO BAG RED RETROSPOT.

### 5. Data quality issues required business-aware treatment
The dataset contained missing CustomerIDs, cancellations, accounting adjustments, zero-priced records and extreme quantities. These records were not automatically deleted. Instead, they were investigated and separated where necessary so that standard sales calculations were based on transactions meeting defined business rules.

## 5. Data Limitations

Several limitations should be considered when interpreting the results:

- Approximately 25.16% of records have a missing CustomerID, limiting customer-level analysis.
- December 2011 contains data only through 9 December and is therefore a partial month.
- Some transactions contain unusually large quantities or values and may influence averages and product rankings.
- The dataset contains cancellations, accounting adjustments and other non-standard records that require separate treatment from standard sales.
- Customer and transaction analysis is based on the available fields and does not provide information about customer demographics, marketing channels or acquisition costs.

## 6. Business Implications

Based on the analysis, the following areas could be considered by the business:

- Investigate the factors contributing to the strong increase in sales and transaction volume between September and November 2011.
- Monitor the UK market closely while exploring opportunities to grow higher-value international markets.
- Develop customer retention strategies for high-value customers and monitor changes in their purchasing behaviour.
- Review high-performing products for potential inventory planning and promotional opportunities.
- Improve data capture processes to reduce missing CustomerID values and improve the reliability of customer-level analysis.
- Maintain separate treatment of cancellations, accounting adjustments and other operational records when calculating sales performance.

## 7. Conclusion

The analysis demonstrates how structured data cleaning and exploratory analysis can reveal meaningful patterns in e-commerce performance.

The business generated approximately 10.63 million in standard sales value during the analysis period, with sales concentrated in the UK and among a smaller group of high-value customers. Sales and transaction volume increased substantially toward the end of the period.

The analysis also demonstrated the importance of data quality validation. Missing customer information, cancellations, accounting adjustments, zero-priced records and extreme transactions required business-aware treatment before drawing conclusions.

Overall, the cleaned dataset provides a stronger foundation for further analysis of customer behaviour, product performance and sales trends.
